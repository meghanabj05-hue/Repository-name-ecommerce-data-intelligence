```python
import psycopg
```

```python
#Connect Python to PostgreSQL
conn = psycopg.connect(
    host="localhost",
    port=5432,
    dbname="ecommerce_db",
    user="postgres",
    password="meghana@123"
)
```

```python
#Test the connection

with conn.cursor() as cur:
    cur.execute("SELECT current_database();")
    print(cur.fetchone())
```

```python
with conn.cursor() as cur:
    cur.execute("SELECT version();")
    print(cur.fetchone())
```

```python
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

connection_url = URL.create(
    drivername="postgresql+psycopg2",
    username="postgres",
    password="meghana@123",
    host="localhost",
    port=5432,
    database="ecommerce_db"
)

engine = create_engine(connection_url)
```

```python
with engine.connect() as conn:
    print("Connected successfully!")
```

```python
# Why are we using Python AND PostgreSQL?

# This is an important distinction.

# We're not replacing Python.

# We're building:

# Python
#   │
#   │ psycopg
#   ▼
# PostgreSQL

# Python will handle things like:

# ETL
# data cleaning
# file loading
# automation
# validation
# ML

# PostgreSQL will handle:

# storage
# relationships
# constraints
# SQL
# aggregation
# analytics

# This is much closer to how real data systems work.
```
